## Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)
```

## Reading the dataset

```python
# Load the dataset
df = pd.read_csv('data/new_retail_data.csv')
# Filter dataset for customers from Germany
df = df[df['Country'] == 'Germany']
# Display first few rows
df.head()
```

## Data Cleaning

```python
# Drop duplicates if any
df = df.drop_duplicates()
```

```python
#Convert numeric columns to integer
cols = ['Transaction_ID','Customer_ID','Zipcode','Age','Year','Total_Purchases']
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')  
df[cols] = df[cols].fillna(0).astype(np.int64)
```

```python
#Format Phone number to a standard form
df['Phone']=df['Phone'].apply(lambda x: str(x))
df['Phone']= df['Phone'].apply(lambda x: x[0:3]+'-'+x[3:6]+'-'+x[6:10])
```

```python
# Convert date column to datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Format the date as day-month-year
df['Date'] = df['Date'].dt.strftime('%d-%m-%Y')
```

```python
# Convert time column to datetime
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M', errors='coerce')

# Format time to show hours and minutes only
df['Time'] = df['Time'].dt.strftime('%H:%M')
```

```python
# Convert columns to numeric, forcing non-convertible values to NaN
df['Age'] = df['Age'].replace('', pd.NA)
df['Total_Amount'] = df['Total_Amount'].replace('', pd.NA)
df['Total_Purchases'] = df['Total_Purchases'].replace('', pd.NA)

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Total_Amount'] = pd.to_numeric(df['Total_Amount'], errors='coerce')
df['Total_Purchases'] = pd.to_numeric(df['Total_Purchases'], errors='coerce')
```

```python
# changing the float numbers to 2 decimal places
pd.set_option('display.float_format', lambda x: '%.2f' % x)
df.head()
```

```python
# filling out missing values
df['Amount'] = df['Amount'].replace('',0)
df['Total_Amount'] = df['Total_Amount'].fillna(df['Total_Purchases']*df['Amount'])
df['Date'] = df['Date'].fillna('')
df['Time'] = df['Time'].fillna('')
```

```python
# Function to generate a random date
def random_date(year, month):
    try:
        # Get the number of days in the given month and year
        days_in_month = pd.Period(f"{year}-{month}").days_in_month
        # Generate a random day
        day = np.random.randint(1, days_in_month + 1)
        return pd.to_datetime(f"{year}-{month}-{day}")
    except:
        return pd.NaT  # Return NaT if the date cannot be created

# Ensure 'Date' column is in datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Apply function only where 'Date' is missing
df.loc[df['Date'].isna(), 'Date'] = df.apply(
    lambda row: random_date(row['Year'], row['Month']) if pd.isna(row['Date']) else row['Date'], axis=1
)
```

```python
# checking rows with null values
df.isnull().sum() 
```

```python
# filling missing product categories
df['Product_Category'] = df.groupby('Product_Type')['Product_Category'].transform(lambda x: x.ffill().bfill())
```
